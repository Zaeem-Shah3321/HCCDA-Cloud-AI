```python
def signup():
    name = input("Enter Name: ")
    pas = input("Enter Password: ")
    print("Signed Up Sucessfully...")
    return name, pas

def login(name, pas):
    name1 = input("Enter Name: ")
    pas1 = input("Enter Password: ")
    if name1 == name and pas1 == pas:
        return True
    else:
        return False

def menu():
    print("1. Signup")
    print("2. Login")
    print("3. Exit")
    opt = int(input("Chose (1, 2, 3): "))
    return opt


if __name__ == "__main__":
    signedIn = False
    while True:
        op = menu()
        if op == 1:
            cred = signup()
            signedIn = True
        elif op == 2:
            if signedIn:
                res = login(*cred)
                if res:
                    print("Login Sucessfull..")
                    break
                else:
                    print("Invalid Name Or Password...")
            else: 
                print("Please Create An Account First...")
        elif op ==3:
            print("Good Bye!")
            break
        else:
            print("Invalid Option...")
```
