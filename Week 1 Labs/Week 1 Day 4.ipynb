```python
def signup():
    name = input("Enter Name: ")
    pas = input("Enter Password: ")
    print("Signed Up Sucessfully...")
    return name, pas

def login(name, pas):
    name1 = input("Enter Name: ")
    pas1 = input("Enter Password: ")
    if name1 == name and pas1 == pas:
        return True
    else:
        return False

def menu():
    print("1. Signup")
    print("2. Login")
    print("3. Exit")
    opt = int(input("Chose (1, 2, 3): "))
    return opt


if __name__ == "__main__":
    signedIn = False
    while True:
        op = menu()
        if op == 1:
            cred = signup()
            signedIn = True
        elif op == 2:
            if signedIn:
                res = login(*cred)
                if res:
                    print("Login Sucessfull..")
                    break
                else:
                    print("Invalid Name Or Password...")
            else:
                print("Please Create An Account First...")
        elif op ==3:
            print("Good Bye!")
            break
        else:
            print("Invalid Option...")
```

```python
def deposit(bln):
    dep = int(input("Enter Amount To Deposit: "))
    bln += dep
    print("Amount Deposited Sucessfully...")
    return bln

def withdraw(balnc):
    wid1 = int(input("Enter Amout to Withdraw: "))
    if wid1 > balnc:
        print("Not Enough Balance...")
    else:
        print("Amount Withdraw Sucessfully...")
        balnc -= wid1
    return balnc

def menu():
    print("1. Deposit")
    print("2. Withdraw")
    print("3. Balance Check")
    print("4. Exit")
    opt = int(input("Chose (1, 2, 3, 4): "))
    return opt


if __name__ == "__main__":
    bal = 100
    while True:
        op = menu()
        if op == 1:
            bal = deposit(bal)
        elif op == 2:
            bal = withdraw(bal)
        elif op == 3:
            print("Balance Is: ", bal)
        elif op == 4:
            print("Good Bye!")
            break
        else:
            print("Invalid Option...")
```

```python
def ad(lst):
    pro = input("Enter Product Name To Add: ")
    lst.append(pro)
    print("Item Added Sucessfully...")

def rm(lst):
    prd = input("Enter Product Name To Remove: ")
    if prd in lst:
        lst.remove(prd)
        print("Item Removed Sucessfully...")
    else:
        print("Item Not Found...")

def viewall(lst):
    if len(lst) == 0:
        print("List Is Empty...")
    else:
        print("Items In The List Are: ")
        for itm in lst:
            print(itm)

def menu():
    print("1. Add Product")
    print("2. Remove Product")
    print("3. View All Products")
    print("4. Exit")
    opt = int(input("Chose (1, 2, 3, 4): "))
    return opt


if __name__ == "__main__":
    prod_list = []
    while True:
        op = menu()
        if op == 1:
            ad(prod_list)
        elif op == 2:
            rm(prod_list)
        elif op == 3:
            viewall(prod_list)
        elif op == 4:
            print("Good Bye!")
            break
        else:
            print("Invalid Option...")
```

```python
weight = float(input("Enter your weight (kg): "))
height = float(input("Enter your height (meters): "))

bmi = weight / (height ** 2)

print("BMI:", round(bmi, 2))

if bmi < 18.5:
    print("Category: Underweight")
elif bmi < 25:
    print("Category: Healthy")
else:
    print("Category: Overweight")
```

```python
quiz = {
    "What is the capital of Pakistan?": {
        "A": "Lahore",
        "B": "Karachi",
        "C": "Islamabad",
        "D": "Peshawar",
        "answer": "C"
    },
    "Which language is used for web development?": {
        "A": "HTML",
        "B": "Python",
        "C": "C++",
        "D": "Java",
        "answer": "A"
    },
    "What is 5 + 3?": {
        "A": "6",
        "B": "7",
        "C": "8",
        "D": "9",
        "answer": "C"
    }
}

score = 0

for question, options in quiz.items():
    print("\n" + question)

    print("A:", options["A"])
    print("B:", options["B"])
    print("C:", options["C"])
    print("D:", options["D"])

    user_answer = input("Enter your answer: ").upper()

    if user_answer == options["answer"]:
        print("Correct!")
        score += 1
    else:
        print("Wrong!")

print("\nYour Score:", score, "/", len(quiz))
```
