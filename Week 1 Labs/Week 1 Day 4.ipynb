```python
tempe = list(map(float, input("Enter Temperatures of 7 Days: ").split()))
if len(tempe) == 7:
    average = sum(tempe) / len(tempe)
    print("Average Temperature:", average)
else:
    print("Please Enter Temperatre of Exactly 7 Days.")
```

```python
def signup():
    name = input("Enter Name: ")
    pas = input("Enter Password: ")
    print("Signed Up Sucessfully...")
    return name, pas

def login(name, pas):
    name1 = input("Enter Name: ")
    pas1 = input("Enter Password: ")
    if name1 == name and pas1 == pas:
        return True
    else:
        return False

def menu():
    print("1. Signup")
    print("2. Login")
    print("3. Exit")
    opt = int(input("Chose (1, 2, 3): "))
    return opt


if __name__ == "__main__":
    signedIn = False
    while True:
        op = menu()
        if op == 1:
            cred = signup()
            signedIn = True
        elif op == 2:
            if signedIn:
                res = login(*cred)
                if res:
                    print("Login Sucessfull..")
                    break
                else:
                    print("Invalid Name Or Password...")
            else: 
                print("Please Create An Account First...")
        elif op ==3:
            print("Good Bye!")
            break
        else:
            print("Invalid Option...")
```

```python
def deposit(bln):
    dep = int(input("Enter Amount To Deposit: "))
    bln += dep
    print("Amount Deposited Sucessfully...")
    return bln

def withdraw(balnc):
    wid1 = int(input("Enter Amout to Withdraw: "))
    if wid1 > balnc:
        print("Not Enough Balance...")
    else:
        print("Amount Withdraw Sucessfully...")
        balnc -= wid1
    return balnc

def menu():
    print("1. Deposit")
    print("2. Withdraw")
    print("3. Balance Check")
    print("4. Exit")
    opt = int(input("Chose (1, 2, 3, 4): "))
    return opt


if __name__ == "__main__":
    bal = 100
    while True:
        op = menu()
        if op == 1:
            bal = deposit(bal)
        elif op == 2:
            bal = withdraw(bal)
        elif op == 3:
            print("Balance Is: ", bal)
        elif op == 4:
            print("Good Bye!")
            break
        else:
            print("Invalid Option...")
```

```python
mrk = int(input("Enter Your Marks: "))
if mrk >= 85 and mrk <= 100:
    print("A Grade for", mrk, "Marks")
elif mrk >= 70 and mrk <= 84:
    print("B Grade for", mrk, "Marks")
elif mrk >= 50 and mrk <= 69:
    print("c Grade for", mrk, "Marks")
elif mrk < 50:
    print("F Grade for", mrk, "Marks")
else:
    print("Invalid Marks...")
```

```python
def ad(lst):
    pro = input("Enter Product Name To Add: ")
    lst.append(pro)
    print("Item Added Sucessfully...")

def rm(lst):
    prd = input("Enter Product Name To Remove: ")
    if prd in lst:
        lst.remove(prd)
        print("Item Removed Sucessfully...")
    else:
        print("Item Not Found...")

def viewall(lst):
    if len(lst) == 0:
        print("List Is Empty...")
    else:
        print("Items In The List Are: ")
        for itm in lst:
            print(itm)

def menu():
    print("1. Add Product")
    print("2. Remove Product")
    print("3. View All Products")
    print("4. Exit")
    opt = int(input("Chose (1, 2, 3, 4): "))
    return opt


if __name__ == "__main__":
    prod_list = []
    while True:
        op = menu()
        if op == 1:
            ad(prod_list)
        elif op == 2:
            rm(prod_list)
        elif op == 3:
            viewall(prod_list)
        elif op == 4:
            print("Good Bye!")
            break
        else:
            print("Invalid Option...")
```
