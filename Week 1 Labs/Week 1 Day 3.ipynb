```python
str1 = input("Write Any Word: ")
str2 = str1[::-1]
if str1 == str2:
  print("String Is Palindrome.")
else:
  print("String Is Not Palindrome.")
```

```python
str2 = input("Type Any String: ")
vow = 0
cons = 0
for char in str2:
  if char in 'aeiouAEIOU':
    vow += 1
  else:
    cons += 1
print("There Are ", vow, "Vowels And ", cons, "Consonents In The String")
```

```python
def factorial(num6):
    if num6 == 0 or num6 ==1:
        return 1
    return num6 * factorial(num6 - 1)

if __name__ == "__main__":
    num7 = int(input("Enter Any Number: "))
    print("Factorial of", num7, "is: " ,factorial(num7))
```

```python
def reverse(num8):
    num9 = num8[::-1]
    return num9

if __name__ == "__main__":
    num8 = list(map(int, input("Enter numbers (Seperated By Sapces): ").split()))
    num10 = reverse(num8)
    print("Reverse of Your List Is: ", num10)
```

```python
num11 = int(input("Enter Any Number: "))
if num11 < 2:
    print("Number is Not Prime")
else:
    for i in range(2,num11):
        if num11 % i == 0:
            print("Number is Not Prime")
            break
    else:
        print("Number is Prime")
```
