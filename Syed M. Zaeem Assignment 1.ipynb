```python
# Import NumPy
import numpy as np
```

```python
# Array Having 10 Zeros
ar1 = np.zeros(10)
ar1
```

```python
# Array on 10 Ones
ar0 = np.ones(10)
ar0
```

```python
# Converting 1D Array Into 2D Array
ar2 = np.arange(1,13)
print("1-D Array: ", ar2)
ar3 = ar2.reshape(3,4)
print("2D Array: ", ar3)
```

```python
# Finding Maximum And Minimum in Array
print("Minimum is: ", ar2.min())
print("Maximum is: ", ar2.max())
```

```python
# Finding Mean
ar2.mean()
```

```python
# Finding Median
ar4 = np.median(ar2)
ar4
```

```python
# Finding Standard Deviation
ar5 = np.std(ar2)
ar5
```

```python
# Identity Matrix of Size 3x3
np.eye(3,3)
```

```python
# Replace All Negative Numbers In Array With 0
ar6 = np.array([1, -9, 6, -2, -7, 8, -4])
ar6[ar6 < 0] = 0
ar6
```

```python
# Filter All Elements Graeter Than 20
ar7 = np.array([7, 9, 21, 13, 43, 88, 90, 54, 24, 9, 1, 0])
ar8 = ar7[ar7 > 20]
ar8
```

```python
# Add And Multiply Two Arrays Element Wise
ar9 = np.arange(1,20,2)
ar10 = np.arange(21,40,2)
print("Sum Is: ", np.add(ar9,ar10))
print("Product Is: ", np.multiply(ar9,ar10))
```

```python
# Concatenate 2 Arrays Vertically
ar11 = np.array([[1,2,3,4,5], [9,8,7,6,5]])
ar12 = np.array([[1,7,9,2,5], [3,8,4,6,5]])
ar13 = np.concatenate((ar11,ar12), axis = 1)
ar13
```
