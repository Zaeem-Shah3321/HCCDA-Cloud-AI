```python
# Import NumPy
import numpy as np
```

```python
# Array Having 10 Zeros
ar1 = np.zeros(10)
ar1
```

```python
# Array on 10 Ones
ar0 = np.ones(10)
ar0
```

```python
# Converting 1D Array Into 2D Array
ar2 = np.arange(1,13)
print("1-D Array: ", ar2)
ar3 = ar2.reshape(3,4)
print("2D Array: ", ar3)
```

```python
# Finding Maximum And Minimum in Array
print("Minimum is: ", ar2.min())
print("Maximum is: ", ar2.max())
```

```python
# Finding Mean
ar2.mean()
```

```python
# Finding Median
ar4 = np.median(ar2)
ar4
```

```python
# Finding Standard Deviation
ar5 = np.std(ar2)
ar5
```

```python
# Identity Matrix of Size 3x3
np.eye(3,3)
```
